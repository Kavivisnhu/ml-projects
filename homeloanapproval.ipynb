```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/loan_data.csv')
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['person_gender']=le.fit_transform(df['person_gender'])
df['person_education']=le.fit_transform(df['person_education'])
df['person_home_ownership']=le.fit_transform(df['person_home_ownership'])
df['loan_intent']=le.fit_transform(df['loan_intent'])
df['previous_loan_defaults_on_file']=le.fit_transform(df['previous_loan_defaults_on_file'])

```

```python
df
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
df['person_age']=sc.fit_transform(df[['person_age']])
df['person_income']=sc.fit_transform(df[['person_income']])
df['person_emp_exp']=sc.fit_transform(df[['person_emp_exp']])
df['loan_amnt']=sc.fit_transform(df[['loan_amnt']])
df['loan_int_rate']=sc.fit_transform(df[['loan_int_rate']])
df['cb_person_cred_hist_length']=sc.fit_transform(df[['cb_person_cred_hist_length']])
df['credit_score']=sc.fit_transform(df[['credit_score']])
```

```python
df
```

```python
x=df.drop('loan_status',axis=1)
y=df['loan_status']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression(solver='liblinear')
rfe=RFE(estimator=model,n_features_to_select=8)
rfe.fit(x_train,y_train)

print('Selected features mask:', rfe.support_)
print('Selected feature names:', x_train.columns[rfe.support_].tolist())
print('Feature ranking:', rfe.ranking_)

# Transform x_train and x_test to include only the selected features
x_train = rfe.transform(x_train)
x_test = rfe.transform(x_test)
```

```python
selected_features = x.columns[rfe.support_]
print('top 8 features this model')
for feature in selected_features:
  print(feature)
```

```python
from imblearn.over_sampling import SMOTE
sm=SMOTE()
```

```python
x_train,y_train=sm.fit_resample(x_train,y_train)
```

```python
y_train.value_counts()
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```
